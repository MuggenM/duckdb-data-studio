```python
import pandas as pd
import requests
```

```python
STUDIO_URL = "http://duckdb-studio:8085"
ENDPOINT = "parameterquery"
PARAMETERS = {"createddate": "2015-01-17"}

response = requests.get(f"{STUDIO_URL}/api/{ENDPOINT}?createddate=Lt2015-01-17")#, params=PARAMETERS)

if response.status_code == 200:
    # Load JSON directly into a Pandas DataFrame
    df = pd.DataFrame(response.json())
    
    # Display the beautiful interactive table
    display(df.head())
else:
    print("Error:", response.text)
```
